```
import pandas as pd

from comp_arch_hw2 import directory
```

```
direct_mapped = pd.read_csv(directory.data / "direct-mapped.csv")
least_recently_used = pd.read_csv(directory.data / "least-recently-used.csv")
round_robin = pd.read_csv(directory.data / "round-robin.csv")
```

```
direct_mapped.head()
```

```
direct_mapped.pivot(
    index="block_size_bytes",
    columns="cache_size_kib",
    values="miss_rate",
).mul(100).style.format("{:.5f}%")
```

```
def associative_pivot(df: pd.DataFrame, label) -> pd.DataFrame:
    return (
        df.pivot(
            index="cache_size_kib",
            columns="associativity",
            values="miss_rate",
        )
        .rename(columns=lambda x: f"{x}-way ({label})")
        .mul(100)
    )


pd.concat(
    [
        associative_pivot(least_recently_used, "LRU"),
        associative_pivot(round_robin, "RR"),
    ],
    axis=1,
).style.format("{:.5f}%")
```
